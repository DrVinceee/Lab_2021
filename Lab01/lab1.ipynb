## Python 数据类型和数据结构

Python 有以下基本数据类型:

- **字符串 (string)**: `"A"`, `'hello'`

Python 的主要数据结构包括:

````python
# Python 列表示例
x = [1, 2, 3, 4, 5]
print("列表:", x)
print("第一个元素 (索引从0开始):", x[0])  # Python索引从0开始

# NumPy 数组 - 向量化操作
x_array = np.array([1, 2, 3, 4, 5])
print("\nNumPy数组平方:", x_array**2)
print("NumPy数组平方根:", np.sqrt(x_array))

# 逻辑索引
print("小于3的元素:", x_array[x_array < 3])
````

## 使用航班数据进行数据操作

我们将使用seaborn内置的flights数据集来演示pandas的数据操作功能。

````python
# 加载示例数据集
# 注意: 这里使用seaborn的flights数据集作为示例
flights = sns.load_dataset('flights')
print("航班数据集前几行:")
print(flights.head())

print("\n数据集形状:", flights.shape)
print("\n列名:", flights.columns.tolist())
````

## 数据过滤 (filter)

选择满足特定条件的行。

````python
# 过滤: 选择1月份的数据
january_flights = flights[flights['month'] == 'Jan']
print("1月份的航班数据:")
print(january_flights.head())

# 多条件过滤
jan_1949 = flights[(flights['month'] == 'Jan') & (flights['year'] == 1949)]
print("\n1949年1月的航班:")
print(jan_1949)
````

## 选择列 (select)

选择特定的列。

````python
# 选择特定列
selected_columns = flights[['year', 'month', 'passengers']]
print("选择的列:")
print(selected_columns.head())
````

## 添加新列 (mutate)

基于现有列创建新列。

````python
# 添加新列
flights_with_new_col = flights.copy()
flights_with_new_col['passengers_thousands'] = flights_with_new_col['passengers'] / 1000
print("添加新列后:")
print(flights_with_new_col.head())
````

## 排序 (arrange)

按照列值对数据进行排序。

````python
# 按乘客数升序排序
sorted_flights = flights.sort_values('passengers')
print("按乘客数升序排序:")
print(sorted_flights.head())

# 按乘客数降序排序
sorted_flights_desc = flights.sort_values('passengers', ascending=False)
print("\n按乘客数降序排序:")
print(sorted_flights_desc.head())
````

## 分组和汇总 (group_by 和 summarize)

对数据进行分组并计算汇总统计量。

````python
# 按年份分组，计算平均乘客数
yearly_avg = flights.groupby('year')['passengers'].mean().reset_index()
yearly_avg.columns = ['year', 'avg_passengers']
print("每年平均乘客数:")
print(yearly_avg)

# 更复杂的汇总: 按年份分组，计算多个统计量
yearly_stats = flights.groupby('year')['passengers'].agg([
    ('count', 'count'),
    ('mean', 'mean'),
    ('min', 'min'),
    ('max', 'max')
]).reset_index()
print("\n每年的详细统计:")
print(yearly_stats)
````

## 方法链式调用

Pandas支持方法链式调用，可以连续执行多个操作。

````python
# 链式调用: 过滤、分组、汇总
result = (flights
    .query('year >= 1950')  # 过滤1950年及以后的数据
    .groupby('month')['passengers']  # 按月份分组
    .mean()  # 计算平均值
    .sort_values(ascending=False)  # 降序排序
    .reset_index()
)
result.columns = ['month', 'avg_passengers']
print("1950年后各月份平均乘客数（降序）:")
print(result)
````

## 数据可视化

使用matplotlib和seaborn进行数据可视化。

````python
# 使用matplotlib绘制时间序列图
plt.figure(figsize=(12, 6))
for month in flights['month'].unique()[:3]:  # 只展示前3个月
    month_data = flights[flights['month'] == month]
    plt.plot(month_data['year'], month_data['passengers'], marker='o', label=month)

plt.xlabel('年份 (Year)', fontsize=12)
plt.ylabel('乘客数 (Passengers)', fontsize=12)
plt.title('航班乘客数随时间变化', fontsize=14)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# 使用seaborn绘制箱线图
plt.figure(figsize=(14, 6))
sns.boxplot(data=flights, x='month', y='passengers', 
            order=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
                   'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.xlabel('月份 (Month)', fontsize=12)
plt.ylabel('乘客数 (Passengers)', fontsize=12)
plt.title('各月份航班乘客数分布', fontsize=14)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

````python
# 热力图展示年份和月份的乘客数
# 透视数据以创建矩阵形式
pivot_flights = flights.pivot_table(values='passengers', 
                                    index='month', 
                                    columns='year',
                                    aggfunc='sum')

# 重新排序月份
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
pivot_flights = pivot_flights.reindex(month_order)

plt.figure(figsize=(12, 8))
sns.heatmap(pivot_flights, cmap='YlOrRd', annot=False, fmt='d', cbar_kws={'label': '乘客数'})
plt.xlabel('年份 (Year)', fontsize=12)
plt.ylabel('月份 (Month)', fontsize=12)
plt.title('航班乘客数热力图', fontsize=14)
plt.tight_layout()
plt.show()
````

## Bash 脚本基础

### 什么是bash?

> Bash是一个命令处理器，通常在文本窗口中运行，用户在其中输入导致操作的命令。Bash也可以从文件中读取和执行命令，称为shell脚本。

### 为什么使用它?

- 当你没有图形用户界面时，即没有图标，没有鼠标...
- 当你想要执行命令100,000次时（或者如果你有100,000个文件要打开），鼠标点击可能太多了。

### Bash命令

一般来说，shell命令由以下部分组成：
> [命令] [选项] [参数]

空格对于解析每个命令至关重要，这与R和Python不同。

### 示例命令:

```bash
# 改变目录
cd /path/to/directory

# 列出目录内容
ls
ls -alh ..

# 统计行数
wc -l states.txt

# 使用管道
cat states.txt | wc -l
```

### 管道 (Pipe)

你可以使用管道 (`|`) 将前一个命令的STDOUT作为后续命令的STDIN。

### 获取帮助

如果你需要特定命令的帮助，使用 `man [command]`。

## Bash 变量

在Bash中，你可以在变量中存储数据。使用等号 (`=`) 将数据分配给变量。

```bash
#!/bin/bash

# 定义变量
lab_number=1
echo $lab_number

the_empire_state="New York"
echo $the_empire_state

# 注意: 等号周围不允许有空格

# 使用let修改变量值
let lab_number=$lab_number+1
echo $lab_number

# 命令替换
state_n=$(cat states.txt | wc -l)
echo "There are $state_n lines in states.txt"
```

## Bash 数学运算

Bash编程语言可以进行非常基本的算术运算。

```bash
#!/bin/bash
expr 5 + 2      # 加法: 7
expr 5 \* 2     # 乘法: 10 (需要转义*)
expr 5 / 2      # 除法: 2 (整数除法)
expr 40 % 21    # 取模: 19
```

## Bash 条件执行

### AND / OR 操作符

```bash
# AND操作符 (&&): 只有左侧命令成功(退出状态为0)时，才执行右侧命令
true && echo "Program 1 was executed."
false && echo "Program 2 was executed."

# OR操作符 (||): 只有左侧命令失败时，才执行右侧命令
true || echo "Program 1 was executed."
false || echo "Program 2 was executed."
```

### 条件表达式

```bash
# 数值比较
[[ 4 -gt 3 ]]  # 大于 (greater than)
echo $?        # 0 (成功)

[[ 3 -gt 4 ]]
echo $?        # 1 (失败)

# 文件测试
[[ -e var.sh ]] && echo "文件存在" || echo "文件不存在"
```

### If-Else 语句

```bash
#!/bin/bash

if [[ $1 -gt $2 ]]; then
    echo "$1 is greater than $2"
elif [[ $1 -eq $2 ]]; then
    echo "$1 equals $2"
else
    echo "$1 is less than $2"
fi
```
