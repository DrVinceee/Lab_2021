# Lab 8: 单细胞RNA测序分析 (scRNA-seq)

本notebook使用Scanpy（Python的Seurat替代品）进行单细胞RNA-seq分析。

**Scanpy vs Seurat对照**:
- Seurat::CreateSeuratObject → scanpy.read_*
- Seurat::NormalizeData → sc.pp.normalize_total + sc.pp.log1p
- Seurat::FindVariableFeatures → sc.pp.highly_variable_genes
- Seurat::RunPCA → sc.tl.pca
- Seurat::FindNeighbors + FindClusters → sc.pp.neighbors + sc.tl.leiden
- Seurat::RunUMAP → sc.tl.umap

```python
# 安装scanpy: pip install scanpy
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 设置scanpy参数
sc.settings.verbosity = 3
sc.settings.set_figure_params(dpi=80, facecolor='white')

print('Scanpy单细胞分析环境准备就绪')
```
